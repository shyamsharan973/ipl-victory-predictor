```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
first_dataset = pd.read_csv('deliveries_updated_ipl_upto_2025.csv')
first_dataset.head(10)
```

```python
second_dataset = pd.read_csv('matches_updated_ipl_upto_2025.csv')
second_dataset
```

```python
total_runs = first_dataset.groupby(['matchId','inning']).sum()['total_run'].add(1).reset_index()
total_runs
```

```python
#Selecting only first innings score to find target of each match
total_runs = total_runs[total_runs['inning']==1]
```

```python
match_df = second_dataset.merge(total_runs[['matchId','total_run']],left_on='matchId',right_on='matchId')
```

```python
match_df
```

```python
match_df = match_df[["matchId","city","winner","total_run"]]
```

```python
match_df
```

```python
match_df = match_df.rename(columns={"matchId" : "match_id"})
```

```python
match_df = match_df.rename(columns={"total_run" : "total_runs"})
```

```python
match_df
```

```python
first_dataset = first_dataset.rename(columns={"matchId" : "match_id"})
```

```python
second_dataset = first_dataset.rename(columns={"matchId" : "match_id"})
```

```python
second_innings =  match_df.merge(first_dataset,on="match_id")
second_innings
```

```python
second_innings = second_innings[second_innings['inning']==2]
second_innings.shape
```

```python
second_innings['batting_team'].value_counts()
```

```python
second_innings['batting_team'] = second_innings['batting_team'].replace('Delhi Daredevils','Delhi Capitals')
second_innings['bowling_team'] = second_innings['bowling_team'].replace('Delhi Daredevils','Delhi Capitals')
```

```python
second_innings['batting_team'] = second_innings['batting_team'].replace('Deccan Chargers','Sunrisers Hyderabad')
second_innings['bowling_team'] = second_innings['bowling_team'].replace('Deccan Chargers','Sunrisers Hyderabad')
```

```python
second_innings['batting_team'].value_counts()
```

```python
second_innings['batting_team'] = second_innings['batting_team'].replace('Punjab Kings','Kings XI Punjab')
second_innings['bowling_team'] = second_innings['bowling_team'].replace('Punjab Kings','Kings XI Punjab')
```

```python
second_innings['batting_team'].value_counts()
```

```python
teams = ['Kolkata Knight Riders', 
         'Chennai Super Kings',
         'Kings XI Punjab', 
         'Rajasthan Royals',
         'Mumbai Indians',
         'Delhi Capitals',
         'Royal Challengers Bangalore',
         'Sunrisers Hyderabad',
        'Lucknow Super Giants',
         'Gujarat Titans']
```

```python
second_innings = second_innings[second_innings['batting_team'].isin(teams)]
second_innings = second_innings[second_innings['bowling_team'].isin(teams)]
```

```python
second_innings['bowling_team'].value_counts()
```

```python
second_innings['batting_team'].value_counts()
```

```python
second_innings['current_Score'] = (
    second_innings
    .groupby('match_id')['total_run']
    .cumsum()
)
second_innings
```

```python
#Simple maths to calculate runs left after each ball
second_innings['runs_left'] = second_innings['total_runs'] - second_innings['current_Score']
second_innings
```

```python
#Calculating ball left after each ball is thrown using unique formula
second_innings['balls_left'] = 120 - (second_innings['over'] * 6 + second_innings['ball'])
second_innings
```

```python
second_innings.dtypes
```

```python
second_innings['player_dismissed'] = second_innings['player_dismissed'].fillna("0")
```

```python
second_innings
```

```python
second_innings['player_dismissed'] = second_innings['player_dismissed'].apply(lambda x:x if x=="0" else "1")
```

```python
second_innings['player_dismissed'] = second_innings['player_dismissed'].astype('int')
```

```python
wickets = (
    second_innings['player_dismissed']
    .notna()
    .groupby(second_innings['match_id'])
    .cumsum()
)
```

```python
wickets = wickets.groupby(second_innings['match_id']).shift(fill_value=0)
```

```python
second_innings['wickets_remaining'] = 10 - wickets
```

```python
second_innings
```

```python
second_innings.drop('wickets_remaining', axis=1, inplace=True)
```

```python
second_innings.head(10)
```

```python
second_innings['player_dismissed'].dtype
```

```python
wickets = second_innings.groupby('match_id')['player_dismissed'].cumsum()
```

```python
second_innings['wickets_remaining'] = 10-wickets
```

```python
second_innings.head(10)
```

Current Run Rate

```python
second_innings['crr'] = round((second_innings['current_Score']*6)/(120 - second_innings['balls_left']),2)
second_innings
```

Required Run Rate

```python
second_innings['rrr'] = round((second_innings['runs_left']*6)/(second_innings['balls_left']),2)
second_innings
```

```python
def winner(team):
    return 1 if team['batting_team'] == team['winner'] else 0
```

```python
second_innings['results'] = second_innings.apply(winner,axis=1)
second_innings.sample()
```

Only Taking required Data

```python
required_data = second_innings[['batting_team','bowling_team','city','runs_left','balls_left','wickets_remaining','total_runs','crr','rrr','results']]
required_data
```

Sampling Data

```python
final_data = required_data.sample(required_data.shape[0])
```

Dropping Null Values

```python
final_data.dropna(inplace=True)
```

Removing rows where balls_left==0 to remove -inf and inf from rrr column

```python
final_data = final_data[final_data['balls_left']!=0]
```

```python
final_data.sample()
```

Storing data in new csv

```python
final_data.to_csv("final_data.csv", index=False)
```

Building Model

```python
from sklearn.model_selection import train_test_split as tts
```

```python
x = final_data.iloc[:, :-1]
y = final_data.iloc[:,-1]
```

```python
x_train,x_test,y_train,y_test = tts(x,y,test_size=0.25,random_state=100)
x_train
```

```python
x_test
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

trf = ColumnTransformer([
    ('trf', OneHotEncoder(sparse_output=False, drop='first'),
     ['batting_team','bowling_team','city'])
],
remainder='passthrough')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ('step1',trf),
    ('step2',LogisticRegression(solver='liblinear'))
])
```

```python
#Training Model
pipe.fit(x_train,y_train)
```

```python
#Testing Model
y_pred = pipe.predict(x_test)
```

```python
#Calculating Accuracy
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
pipe.predict_proba(x_test)[0]
```

```python
import pickle
pickle.dump(pipe,open('pipe.pkl','wb'))
```

Random Forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
trf = ColumnTransformer([
    ('trf',OneHotEncoder(sparse_output=False,drop='first'),['batting_team','bowling_team','city'])
],
remainder='passthrough')

pipe = Pipeline([
    ('step1',trf),
    ('step2',RandomForestClassifier())
])

pipe.fit(x_train,y_train)

y_pred = pipe.predict(x_test)

accuracy_score(y_test,y_pred)
```
